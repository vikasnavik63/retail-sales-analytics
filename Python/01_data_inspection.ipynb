```python
# Test a parser that handles the mixed date formats

def parse_mixed_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    formats = [
        "%Y-%m-%d",   # 2024-09-05
        "%d/%m/%Y",   # 10/09/2024
        "%m/%d/%Y",   # 03/26/2024
        "%d.%m.%Y",   # 10.04.2024
        "%d-%b-%Y",   # 15-Dec-2024
        "%d-%B-%Y"    # 15-December-2024
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue

    return pd.NaT


# Test without changing the original column
date_test = df["Order_Date"].apply(parse_mixed_date)

print("Total rows:", len(df))
print("Successfully parsed:", date_test.notna().sum())
print("Failed to parse:", date_test.isna().sum())

# Show any values that still cannot be parsed
failed_dates = df.loc[date_test.isna(), "Order_Date"].drop_duplicates()

print("\nUnparsed date examples:")
display(failed_dates.head(30))
```

```python
# Calculate fallback Age using Country + Gender + Customer_Segment
# for the 48 remaining records

remaining_age_mask = df["Age"].isna()

fallback_age = (
    df[df["Age"].notna()]
    .groupby(["Country", "Gender", "Customer_Segment"])["Age"]
    .median()
)

remaining_48 = df.loc[
    remaining_age_mask,
    [
        "Customer_ID",
        "Customer_Name",
        "Gender",
        "Country",
        "City",
        "Customer_Segment",
        "Age"
    ]
].copy()

remaining_48["Fallback_Age"] = remaining_48.apply(
    lambda row: fallback_age.get(
        (row["Country"], row["Gender"], row["Customer_Segment"])
    ),
    axis=1
)

print("Remaining records:", len(remaining_48))
print("Records with fallback Age:", remaining_48["Fallback_Age"].notna().sum())
print("Records still without fallback Age:", remaining_48["Fallback_Age"].isna().sum())

display(remaining_48)
```

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_csv("../cleaned_data/retail_sales_clean.csv")

# Convert dates
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"])

# Display basic information
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
# RFM Customer Analysis

analysis_date = df["Order_Date"].max() + pd.Timedelta(days=1)

rfm = (
    df.groupby("Customer_ID")
      .agg(
          Recency=("Order_Date", lambda x: (analysis_date - x.max()).days),
          Frequency=("Order_ID", "nunique"),
          Monetary=("Revenue", "sum")
      )
      .reset_index()
)

rfm["Monetary"] = rfm["Monetary"].round(2)

rfm.head(10)
```

```python
# Create RFM scores

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head(10)
```

```python
# Combined RFM Score

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

rfm.head(10)
```

```python
# Create customer segments

def segment_customer(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "High Value"
    elif row["R_Score"] >= 4:
        return "Recent Customers"
    elif row["F_Score"] >= 4:
        return "Frequent Customers"
    elif row["M_Score"] >= 4:
        return "High Spenders"
    else:
        return "Other"


rfm["Segment"] = rfm.apply(segment_customer, axis=1)

rfm.head(10)
```

```python
# Customer count by RFM segment

segment_counts = (
    rfm["Segment"]
    .value_counts()
    .reset_index()
)

segment_counts.columns = ["Segment", "Customer_Count"]

segment_counts
```

```python
# Revenue and average customer value by RFM segment

segment_revenue = (
    rfm.groupby("Segment")
       .agg(
           Total_Revenue=("Monetary", "sum"),
           Average_Revenue_Per_Customer=("Monetary", "mean"),
           Customer_Count=("Customer_ID", "count")
       )
       .reset_index()
)

segment_revenue["Total_Revenue"] = segment_revenue["Total_Revenue"].round(2)
segment_revenue["Average_Revenue_Per_Customer"] = (
    segment_revenue["Average_Revenue_Per_Customer"].round(2)
)

segment_revenue.sort_values(
    "Total_Revenue",
    ascending=False
)
```

```python
# RFM Segment Revenue Chart

rfm_chart = segment_revenue.sort_values(
    "Total_Revenue",
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    rfm_chart["Segment"],
    rfm_chart["Total_Revenue"]
)

plt.title("Revenue by RFM Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Revenue")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

plt.show()
```

```python
# Sales Channel Performance

channel_analysis = (
    df.groupby("Sales_Channel")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Total_Orders=("Order_ID", "nunique"),
          Total_Quantity=("Quantity", "sum")
      )
      .reset_index()
)

channel_analysis["Average_Order_Value"] = (
    channel_analysis["Total_Revenue"]
    / channel_analysis["Total_Orders"]
).round(2)

channel_analysis["Total_Revenue"] = (
    channel_analysis["Total_Revenue"].round(2)
)

channel_analysis
```
